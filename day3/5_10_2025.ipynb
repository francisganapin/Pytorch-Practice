```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('data.csv')
```

```python
data_np = df[['Age','Salary']].to_numpy()
```

```python
print(data_np)
```

```python
normalized = (data_np - data_np.mean(axis=0)) / data_np.std(axis=0)
```

```python
mean_val = np.mean(df[['Age','Salary']],axis=0)
std_vals = np.std(df[['Age','Salary']],axis=0)
```

```python
print(mean_val)
print(std_vals)
```

```python
df['Salary'] = np.where(df['Salary'] > 55000,55000,df['Salary'])
```

```python
print(df)
```

```python
def custom_score(x):
    return np.sqrt(x['Age'] * x['Salary'])

df['Score'] = df.apply(custom_score,axis=1)
```

```python
print(df)
```

```python
features = df[['Age','Salary']].to_numpy()
weights = np.array([[0.1],[0.0005]])
```

```python
prediction = features @ weights
df['Prediction'] = prediction
```

```python
print(df)
```

```python
diff = df[['Age','Salary']].to_numpy() - np.array([30,60000])
df[['Age_diff','Salary_diff']] = diff
```

```python
print(df)
```

```python
mask =(df['Age'] > 27) & (df['Salary'] < 65000)
filtered_df = df[mask]
```
