```python
orders = np.array([
    (201, 'Burger', 2, 150.0, 'Served'),
    (202, 'Pizza', 1, 300.0, 'Preparing'),
    (203, 'Pasta', 3, 200.0, 'Served'),
    (204, 'Salad', 1, 120.0, 'Cancelled'),
    (205, 'Fries', 4, 80.0, 'Preparing')
], dtype=[
    ('order_id', 'i4'),
    ('item', 'U20'),
    ('quantity', 'i4'),
    ('price_per_item', 'f4'),
    ('status', 'U10')
])
```

```python
print(pd.DataFrame(orders))
```

```python
preparing_orders = orders[orders['status']=='Preparing']
print(pd.DataFrame(preparing_orders))
```

```python
total_per_order = orders['quantity'] * orders['price_per_item']
```

```python
print(total_per_order)
```

```python
df = pd.DataFrame(orders)
df['total_price'] = df['quantity'] * df['price_per_item']
```

```python
print(df[['order_id','item','total_price','status']])
```
