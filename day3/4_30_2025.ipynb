```python
import pandas as pd
import numpy as np




data = {
    'Department':['Sales','Sales','HR','HR','IT','IT'],
    'Employee':['Alice','Bob','Charlie','David','Eve','Frank'],
    'Salary': [50000, 60000, 70000, 80000, 90000, 100000],
    'Bonus': [5000, 7000, 8000, 9000, 12000, 15000]

}
```

```python
df = pd.DataFrame(data)
```

```python
agg_funcs = {
    'Salary': ['sum', 'mean', 'max'],
    'Bonus': ['mean', lambda x: np.percentile(x, 75)]
}

# Applying Aggregation
result = df.groupby('Department').agg(agg_funcs)
```

```python
result.columns = ['Salary Sum', 'Salary Mean', 'Salary Max', 'Bonus Mean', 'Bonus 75th Percentile']

print(result)
```

```python
import pandas as pd


data = {
    "Store":['A','A','A','B','B','C','C'],
    "Category":["Electronics","Clothing","Clothing","Electronics","Clothing","Electronics","Clothing"],
    "Day":["Monday","Monday","Tuesday","Monday","Tuesday","Monday","Tuesday"],
    "Revenue":[5000,2000,2500,7000,3000,6000,4000]
}
```

```python
df = pd.DataFrame(data)

df.set_index(['Store','Category','Day'],inplace=True)

print(df)
```

```python
grouped = df.groupby(level="Store").sum()

print(grouped)
```

```python
clothing_revenue = df.xs('Clothing',level='Category')
print(clothing_revenue)
```

```python
agg_result = df.groupby(level=["Store",'Category']).agg(
    Total_Revenue =('Revenue','sum'),
    Avg_Revenue=('Revenue','mean')
)

print(agg_result)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Store": ["A", "A", "B", "B", "C", "C", "A", "C", "B"],
    "Product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop", "Phone", "Tablet", "Tablet", "Phone"]
})

# Crosstab - Count occurrences of each product per store
cross_tab_df = pd.crosstab(df["Store"], df["Product"])

print(cross_tab_df)
```

```python
df = pd.DataFrame({
    "Store": ["A", "A", "B", "B", "C", "C", "A", "C", "B"],
    "Product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop", "Phone", "Tablet", "Tablet", "Phone"]
})
```

```python
cross_tab_margin = pd.crosstab(df["Store"],df["Product"],margins=True)

print(cross_tab_margin)
```

```python
normalized_df = pd.crosstab(df["Store"],df["Product"],normalize='index',margins=True)
print(normalized_df)
```
