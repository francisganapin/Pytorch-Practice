```python
import json

data = '''
{
    "students": [
        {"id": 1, "name": "Alice", "scores": {"math": 90, "science": 85}},
        {"id": 2, "name": "Bob", "scores": {"math": 75, "science": 95}},
        {"id": 3, "name": "Charlie", "scores": {"math": 88, "science": 72}}
    ]
}
'''

parsed = json.loads(data)
```

```python
for student in parsed['students']:
    print(student['name'], "->",student['scores'])
```

```python
#show all math scores
math_scores =[s['scores']['math'] for s in parsed['students']]
print(math_scores)
```

```python
for student in parsed['students']:
    if student['scores']['science'] > 90:
        print(student['name'])
```

```python
import json
from collections import Counter
```

```python
data = '''
{
    "orders": [
        {"customer": "Alice", "item": "Whey Protein"},
        {"customer": "Bob", "item": "Creatine"},
        {"customer": "Alice", "item": "Whey Protein"},
        {"customer": "Charlie", "item": "Mass Gainer"},
        {"customer": "Bob", "item": "Whey Protein"}
    ]
}
'''
```

```python
parsed = json.loads(data)

```

```python
items = [order['item'] for order in parsed['orders']]
```

```python
print(items)
```

```python
item_counter = Counter(items)
```

```python
print(item_counter)
```

```python
print(item_counter.most_common())
```

```python
json_data = '''
{
    "orders": [
        {"item": "Whey Protein"},
        {"item": "Creatine"},
        {"item": "Whey Protein"},
        {"item": "Mass Gainer"},
        {"item": "Whey Protein"},
        {"item": "Creatine"},
        {"item": "Pre-Workout"}
    ]
}
'''
```

```python
data = json.loads(json_data)
items = [order['item'] for order in data['orders']]
```

```python
count = Counter(items)
top_items = count.most_common(2)
```

```python
print(f'top 2 item {top_items}')
```

```python
data = '''
[
  {"category": "supplement", "item": "Whey"},
  {"category": "supplement", "item": "Creatine"},
  {"category": "snack", "item": "Protein Bar"},
  {"category": "supplement", "item": "Mass Gainer"},
  {"category": "snack", "item": "Peanut Butter"}
]
'''
```

```python
parsed = json.loads(data)
grouped = defaultdict(list)
```

```python
from itertools import groupby
```

```python
data = [
    {'customer':'Alice','item':'Whey'},
    {'customer':'Alice','item':'Creatine'},
    {'customer':'Bob','item':'Mass Gainer'},
    {'customer':'Bob','item':'Pre-Workout'}
]

data.sort(key=lambda x: x['customer'])

for key,group in groupby(data,key=lambda x: x['customer']):
    print(key,list(group))
```

```python
data = [
    {'category':'supplement','item':'Whey'},
    {'category':'supplement','item':'Creatine'},
    {'category':'snack','item':'Protein Bar'},
]

grouped = {}

for entry in data:
    key = entry['category']
    grouped.setdefault(key,[]).append(entry['item'])


print(grouped)
```
