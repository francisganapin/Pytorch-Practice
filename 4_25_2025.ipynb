```python
while True:
    num = input("Enter a number (or 'exist' to quit): ")
    if num.lower() == "exist":
        print("Existing ...")
        break
    elif num.isdigit():
        print(f"your entered: {num}")
    else:
        print('Invalid input, try again.')    
```
