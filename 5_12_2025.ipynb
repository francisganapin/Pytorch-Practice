```python
import pandas as pd

data = {
    'OrderID': [1001, 1002, 1003, 1004, 1005, 1006],
    'Customer': ['Alice', 'Bob', 'Alice', 'David', 'Bob', 'Alice'],
    'Region': ['North', 'South', 'North', 'East', 'South', 'North'],
    'Product': ['Laptop', 'Tablet', 'Phone', 'Laptop', 'Phone', 'Tablet'],
    'Quantity': [1, 3, 2, 1, 1, 4],
    'UnitPrice': [1200, 400, 600, 1200, 600, 400],
    'Discount': [0.1, 0.0, 0.05, 0.1, 0.2, 0.0],
    'Date': pd.to_datetime(['2024-01-15', '2024-01-16', '2024-01-17', '2024-01-18', '2024-01-19', '2024-01-20'])
}

df = pd.DataFrame(data)
```

```python
#Total revenue per customer (consider quantity × unit price × (1 - discount))

df['Revenue'] = df['Quantity'] * df['UnitPrice'] * (1- df['Discount'] )
revenue_per_customer = df.groupby('Customer')['Revenue'].sum()
print(revenue_per_customer)
```

```python
popular_product = df.groupby('Product')['Quantity'].sum()
print(pd.DataFrame(popular_product))

```

```python
average_discount = df.groupby('Region')['Discount'].sum()
```

```python
print(pd.DataFrame(average_discount))
```

```python
#Total revenue per product in the North region
```

```python
north_df = df[df['Region']=='North']
```

```python
print(north_df)
print(f"total revenue of North: {north_df['Revenue'].sum()}")
```

```python
import pandas as pd

data = {
    'Customer': ['Anna', 'Ben', 'Cathy', 'David', 'Ella'],
    'Item': ['Milk', 'Bread', 'Eggs', 'Milk', 'Bread'],
    'Quantity': [2, 1, 12, 1, 2],
    'Price_per_item': [1.50, 2.00, 0.20, 1.50, 2.00]
}

df = pd.DataFrame(data)
```

```python
df['Total_Cost'] = df['Quantity'] * df['Price_per_item']
```

```python
print(df)
```

```python
print(f"total cost of all customer: {df['Total_Cost'].sum()}")
```

```python
total_milk = df[df['Item']=='Milk']['Quantity'].sum()
```

```python
print(total_milk)
```

```python
#most spend the most

most_spend = df.loc[df['Total_Cost'].idxmax(),'Customer']
```
