```python
df = pd.DataFrame({
    'price':[100,200,300],
    'discount':[10,20,30]
})

df['final'] = df['price'] - df['discount']
```

```python
print(df)
```

```python
df['tax'] = df['price'] * 0.05
```

```python
df['total'] = df[['price','tax']].sum(axis=1)
```

```python
print(df)
```

```python
df[df['price'] > 150]
```

```python
mask = df['price'] > 100 & (df['discount'] < 25)
```

```python
filtered = df[mask]
```

```python
print(filtered)
```

```python
df['status'] = np.where(df['price'] > 150, 'expensive','cheap')
```

```python
print(df)
```

```python
conditions = [
    df['price'] < 150,
    df['price'].between(150,250),
    df['price'] > 250
]

choices = ['low','medium','high']
df['tier'] = np.select(conditions,choices,default=)
```
