```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python
categories =['Electronics','Clothing','Home','Toys']
```

```python
products = {
    "Electronics": ["Phone", "Laptop", "Headphones"],
    "Clothing": ["T-Shirt", "Jeans", "Jacket"],
    "Home": ["Lamp", "Chair", "Table"],
    "Toys": ["Doll", "Puzzle", "Car"]
}
```

```python
data = []
```

```python
for i in range(200):
    category = np.random.choice(categories)
    product = np.random.choice(products[category])
    price = round(np.random.uniform(10,1000),2)
    quantity = np.random.randint(1,5)
    discount = np.random.choice([0,0.5,0.10,0.20])
    date = pd.to_datetime('2025-05-01') + pd.to_timedelta(np.random.randint(0,30),unit='D')

    data.append([i+1,product,category,price,quantity,discount,date])
```

```python
df = pd.DataFrame(data, columns=["order_id", "product", "category", "price", "quantity", "discount", "date"])
```

```python
print(df)
```

```python
df['revenue'] = df['price'] * df['quantity'] *(1- df['discount'])
```

```python
print(df.head())
```

```python
sns.barplot(data=df, x='category',y='revenue',estimator=sum)
```

```python
daily_revenue = df.groupby('date')['revenue'].sum().reset_index()
```

```python
print(daily_revenue)
```

```python
sns.lineplot(data=daily_revenue,x='date',y='revenue',marker='o')
plt.title('Daily Revenue Trend')
plt.ylabel('Revenue ($)')
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxenplot(data=df,x='discount',y='revenue')
plt.title('Effect of discount on Revenue')
plt.xlabel('Discount (%)')
plt.ylabel('Order Revenue ($)')
plt.show()
```

```python
sns.barplot(data=df,x='category',y='quantity',estimator=sum)
plt.title('Total Quantity sold per Category')
plt.ylabel('Total Item')
plt.show()
```

```python
corr = df[['price','quantity','discount','revenue']].corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Sale Data Correlation')
plt.show()
```
