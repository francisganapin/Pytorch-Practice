```python
#Categorize EVs Based on Range

def range_category(r):
    if r >= 300:
        return 'Long Range'
    elif r >= 150:
        return 'Mid Range'
    else:
        return 'Short Rage'
    
df['Range Category'] = df['Electric Range'].apply(range_category)
df['Range Category'].value_counts()
```
