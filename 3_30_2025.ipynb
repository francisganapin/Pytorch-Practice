```python
arr = np.array([39,40,41,42,43,44,45])

filter_arr = []
filter_arr2 = []

for element in arr:
    if element >= 42:
        filter_arr.append(True)
    elif element < 42:
        filter_arr.append(False)
        filter_arr2.append(True)
    else:
        filter_arr.append(False)
        filter_arr2.append(False)

newarr = arr[filter_arr]
newarr2 = arr[filter_arr2]

print(filter_arr)
print(f' this one is more than or equal 42:{newarr}')
print(f' this one is lower than of 42:{newarr2}')
```

```python
arr = np.array([39, 40, 41, 42, 43, 44, 45])

filter_arr = arr >= 42
filter_arr2 = arr < 42

newarr = arr[filter_arr]
newarr2 = arr[filter_arr2]

print(filter_arr)

print(f'this one is more than or equal to 42: {newarr}')
print(f'this one is less than to 42:{newarr2}')
```
